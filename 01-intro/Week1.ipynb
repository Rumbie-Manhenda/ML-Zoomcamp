```python
# The maximum fuel efficiency of cars from Asia
df[df["origin"]=='Asia' ]["fuel_efficiency_mpg"].max()
```

```python
median = df["horsepower"].describe()['50%']
print(f"The median is {median}")
```

```python
freq_value= df['horsepower'].value_counts(ascending= False).reset_index().max()['horsepower']

print(f"The frequent value in horsepower is {freq_value}")
```

```python
#Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.

null_values = df['horsepower'].isnull().sum() #np.int64(877) null values
print(f"Null values: {null_values}")
df.loc[df['horsepower'].isnull(), 'horsepower'] = freq_value
null_values # check if all null values are gone
```

```python
#Recalculaitng the median
recalculated_median= df['horsepower'].describe()['50%']
print(f"The re-calculated median is {recalculated_median}")
```

```python
all_cars_from_Asia = df.loc[df.origin== 'Asia' , ['model_year', 'vehicle_weight']]
all_cars_from_Asia.head(7)
```

```python
X= np.array(all_cars_from_Asia.head(7))

print(X)
```

```python
XTX= X.T @ X
```

```python
#Invert
xtx_inverse= np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w= xtx_inverse @ X.T @ y

print(w.sum())
```
